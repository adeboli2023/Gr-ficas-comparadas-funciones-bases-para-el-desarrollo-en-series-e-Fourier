```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

def graficar(T=4, guardar=False):
    fig, axes = plt.subplots(2, 2, figsize=(14,8))
    
    # --- Coseno clásico ---
    axes[0,0].plot(x, np.cos(x), label=r"$c_1(x)$", color="blue")
    axes[0,0].plot(x, np.cos(2*x), label=r"$c_2(x)$", color="cyan")
    axes[0,0].set_title("Cosenos clásicos")
    axes[0,0].legend()
    axes[0,0].grid(True)
    
    # --- Coseno con período T ---
    axes[0,1].plot(x, np.cos(2*np.pi*1*x/T), label=fr"$c^T_1(x)$", color="blue")
    axes[0,1].plot(x, np.cos(2*np.pi*2*x/T), label=fr"$c^T_2(x)$", color="cyan")
    axes[0,1].set_title(fr"Cosenos con período $T={T}$")
    axes[0,1].legend()
    axes[0,1].grid(True)
    
    # --- Senos clásicos ---
    axes[1,0].plot(x, np.sin(x), label=r"$s_1(x)$", color="red")
    axes[1,0].plot(x, np.sin(2*x), label=r"$s_2(x)$", color="orange")
    axes[1,0].set_title("Senos clásicos")
    axes[1,0].legend()
    axes[1,0].grid(True)
    
    # --- Senos con período T ---
    axes[1,1].plot(x, np.sin(2*np.pi*1*x/T), label=fr"$s^T_1(x)$", color="red")
    axes[1,1].plot(x, np.sin(2*np.pi*2*x/T), label=fr"$s^T_2(x)$", color="orange")
    axes[1,1].set_title(fr"Senos con período $T={T}$")
    axes[1,1].legend()
    axes[1,1].grid(True)
    
    plt.tight_layout()
    
    # Guardar si el usuario lo pide
    if guardar:
        plt.savefig(f"trigonometria_T{T}.png", dpi=300)
        plt.savefig(f"trigonometria_T{T}.pdf")
    
    plt.show()

# Interfaz interactiva con opción de guardar
interact(graficar, 
         T=widgets.FloatSlider(value=4, min=1, max=10, step=0.5, description="T"),
         guardar=widgets.Checkbox(value=False, description="Guardar imagen"));
```
